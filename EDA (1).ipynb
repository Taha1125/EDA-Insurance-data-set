```python

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
col = ['age', 'bmi', 'children']
df[col] = scaler.fit_transform(df[col])
```

```python
cat_features = [
    'gender', 'smoker', 'children',
    'region_northeast', 'region_northwest',
    'region_southeast', 'region_southwest'
]

for col in cat_features:
    plt.figure(figsize=(6, 2.5))
    sns.countplot(x=df[col])
    plt.title(f'Count of {col}')
    plt.show()
```

```python
# Compare average/distribution of charges across smoker and children
for col in ['gender', 'smoker', 'children']:
    plt.figure(figsize=(6, 3))
    sns.boxplot(x=df[col], y=df['charges'])
    plt.title(f'Charges by {col}')
    plt.show()

# Correlation heatmap across all numeric/encoded columns
plt.figure(figsize=(9, 6))
sns.heatmap(df.corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Feature Correlation Matrix')
plt.show()
```

```python
from scipy.stats import pearsonr
import pandas as pd

alpha = 0.05
num_features = ['age', 'bmi', 'children']
num_results = {}

for col in num_features:
    # Using 'df' instead of 'df_cleaned'
    r_stat, p_val = pearsonr(df[col], df['charges'])
    decision = 'Reject Null (Keep Feature)' if p_val < alpha else 'Accept Null (Drop Feature)'
    num_results[col] = {
        'pearson_r': r_stat,
        'p_value': p_val,
        'Decision': decision
    }

num_df = pd.DataFrame(num_results).T.sort_values(by='p_value')
num_df
```
